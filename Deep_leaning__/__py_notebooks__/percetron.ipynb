```python
import  numpy as np 
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0,0,0,1])
w = np.zeros(X.shape[1])
b = 0.0
lerning_rate = 1.0

for epoch  in  range(50): 
    for x , target in  zip(X,y):
        z = np.dot(x, w) + b
        prediction = 1 if z >= 0 else 0
        error =  target - prediction

        w = w + lerning_rate * error * x
        b = b + lerning_rate * error
        print(z)
    

    print("--------------------")
    print(
        "Epoch:", epoch + 1,
        "Weights:", w,
        "Bias:", b
    )
        
```
