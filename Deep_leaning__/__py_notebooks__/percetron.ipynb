```python
import numpy as np

class Perceptron:

    def __init__(self, learning_rate=1.0, epochs=10):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = 0.0

    def fit(self, X, y):

        self.weights = np.zeros(X.shape[1])
        self.bias = 0.0

        for epoch in range(self.epochs):

            for x, target in zip(X, y):

                z = np.dot(x, self.weights) + self.bias

                prediction = 1 if z >= 0 else 0

                error = target - prediction

                self.weights += self.lr * error * x
                self.bias += self.lr * error

        return self

    def predict(self, X):

        z = X @ self.weights + self.bias

        return (z >= 0).astype(int)
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([0, 0, 0, 1])
model = Perceptron(
    learning_rate=1.0,
    epochs=10
)
model.fit(X, y)
print("Weights:", model.weights)
print("Bias:", model.bias)
print("Predictions:", model.predict(X))
```

```python
import numpy as np

class MulticlassPerceptron:

    def __init__(self, learning_rate=1.0, epochs=10):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def fit(self, X, y):

        n_samples, n_features = X.shape
        n_classes = len(np.unique(y))

        self.weights = np.zeros((n_classes, n_features))
        self.bias = np.zeros(n_classes)

        for epoch in range(self.epochs):

            for x, target in zip(X, y):

                z = self.weights @ x + self.bias

                prediction = np.argmax(z)

                if prediction != target:

                    self.weights[target] += self.learning_rate * x
                    self.bias[target] += self.learning_rate

                    self.weights[prediction] -= self.learning_rate * x
                    self.bias[prediction] -= self.learning_rate

        return self

    def predict(self, X):

        z = X @ self.weights.T + self.bias

        return np.argmax(z, axis=1)

    def score(self, X, y):

        predictions = self.predict(X)

        return np.mean(predictions == y)
```

```python
X = np.array([
    [2, 1],
    [1, 3],
    [3, 2]
], dtype=float)

y = np.array([0, 1, 2])

model = MulticlassPerceptron(
    learning_rate=1.0,
    epochs=10
)
model.fit(X, y)
predictions = model.predict(X)

print(predictions)
```

```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder, StandardScaler
from  sklearn.model_selection import train_test_split

label_encoders = LabelEncoder()
scaler = StandardScaler()

df = pd.read_csv('../__train_dataset__/Data_fruit.csv')
X = df.drop("Class", axis=1)
y = df['Class']

y_encoder = label_encoders.fit_transform(y)

X_train,  X_test ,y_train, y_test = train_test_split(X, y_encoder , test_size=0.2,  random_state=42)

X_train_scaler = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)

model = MulticlassPerceptron(learning_rate=1.0,  epochs=100)

model = MulticlassPerceptron(
    learning_rate=1.0,
    epochs=100
)

model.fit(X_train_scaler, y_train)

print("Weights shape:", model.weights.shape)
print("Bias shape:", model.bias.shape)

predictions = model.predict(X_test_scaler)

print("Predictions:", predictions)

accuracy = model.score(X_test_scaler, y_test)

print("Accuracy:", accuracy)
print("Accuracy:", accuracy * 100, "%")
```
