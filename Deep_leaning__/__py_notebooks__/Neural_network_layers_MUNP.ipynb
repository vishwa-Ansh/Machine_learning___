```python
import numpy as np
X = np.array([2, 3, 1])
W1 = np.array([
    [0.5, 0.2, 0.1],
    [0.1, 0.4, 0.3],
    [0.2, 0.3, 0.6],
    [0.7, 0.1, 0.2]
])
b1 = np.array([0.5, 0.2, -0.1, 0.3])
W2 = np.array([
    [0.2, 0.4, 0.1, 0.3],
    [0.5, 0.1, 0.6, 0.2]
])
b2 = np.array([0.1, -0.2])
Z1 = W1 @ X + b1
A1 = np.maximum(0, Z1)
Z2 = W2 @ A1 + b2
A2 = np.maximum(0, Z2)
print("Z1 =", Z1)
print("A1 =", A1)
print("Z2 =", Z2)
print("A2 =", A2)
```
